```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Flight_Dataset.csv')
df.head()
```

#univariate
histplot with kde

```python
sns.histplot(data=df,x='price',kde=True)
plt.show()
```

##boxplot

```python
sns.boxplot(data=df,x='price')
plt.show()
```

#piechart

```python
df['airline'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title("Pie chart of airlines")
plt.show()
```

#Bivariate
scatterplot

```python
sns.scatterplot(data=df,x='price',y="duration")
plt.show()
```

#Regressionplot

```python
sns.regplot(data=df,x='price',y="duration")
plt.show()
```

#barplot

```python
sns.barplot(data=df,x='airline',y='duration')
plt.show()
```

#countplot with hue

```python
sns.countplot(data=df,x='source_city',hue='destination_city')
plt.show()
```

#multivariate
Pairplot

```python
sns.pairplot(df[['stops','duration','days_left','price']])
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="vlag")
plt.show()
```
